```python
#kmeans
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

dataset = pd.read_csv('./data/processed.csv')
dataset.head()
```

```python
X = dataset.drop('Attrition', axis=1)
y = dataset['Attrition']
```

```python
kmeans = KMeans(n_clusters=2)
kmeans.fit(X)
```

```python
#metrics
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

silhouette = silhouette_score(X, kmeans.labels_)
davies_bouldin = davies_bouldin_score(X, kmeans.labels_)
calinski_harabasz = calinski_harabasz_score(X, kmeans.labels_)

print('silhouette:', silhouette)
print('davies_bouldin:', davies_bouldin)
print('calinski_harabasz:', calinski_harabasz)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
X_t = pca.fit_transform(X)
```

```python
kmeans = KMeans(n_clusters=2)
kmeans.fit(X_t)
```

```python
# plot clusters
plt.scatter(X_t[:, 0], X_t[:, 1], c=kmeans.labels_, cmap='rainbow')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='black')
plt.title('K-means Clustering')
plt.show()
```

```python
silouette = silhouette_score(X_t, kmeans.labels_)

print('silhouetter score:', silouette)

davies_bouldin = davies_bouldin_score(X_t, kmeans.labels_)
print('davies bouldin score:', davies_bouldin)

calinski_harabasz = calinski_harabasz_score(X_t, kmeans.labels_)
print('calinski harabasz score:', calinski_harabasz)
```
